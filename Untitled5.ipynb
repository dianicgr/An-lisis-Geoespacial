```python
# =============================================
# 📦 INSTALACIÓN DE PAQUETES
# =============================================
!pip install geopandas libpysal pymc arviz

# =============================================
# 📁 MONTAJE DE GOOGLE DRIVE
# =============================================
from google.colab import drive
drive.mount('/content/drive')

# =============================================
# 📄 CARGAR GEOBASE DE DATOS
# =============================================
import geopandas as gpd

# Ruta a tu archivo .gpkg
ruta = "/content/drive/MyDrive/Analisis_Geoespacial/entrega_final/Shape/municipios_unificado.gpkg"
gdf = gpd.read_file(ruta, layer="municipios_unificado")  # Cambia "municipios" si tu capa tiene otro nombre

# Revisar columnas
print(gdf.columns)

# =============================================
# ⚙️ CREAR MATRIZ DE VECINOS (ADYACENCIA)
# =============================================
from libpysal.weights import Queen
import numpy as np

# Crear pesos espaciales (vecindad por contigüidad tipo Queen)
w = Queen.from_dataframe(gdf)
W_matrix = w.full()[0]  # Matriz de adyacencia

# =============================================
# 🎯 PREPARAR DATOS PARA EL MODELO
# =============================================
# Variables explicativas (ajusta nombres si es necesario)
X = gdf[['alt_prom']].values
y = gdf['uso_plaguic'].values

n = len(y)

# =============================================
# 🔮 AJUSTAR MODELO CAR BAYESIANO (simplificado)
# =============================================
import pymc as pm
import aesara.tensor as at
import arviz as az

with pm.Model() as car_model:
    # Coeficientes de regresión
    beta = pm.Normal("beta", mu=0, sigma=10, shape=X.shape[1])

    # Parámetro de dispersión espacial
    tau = pm.HalfNormal("tau", sigma=1)

    # Matriz de precisión Q = D - W
    D = np.diag(W_matrix.sum(axis=1))
    Q = D - W_matrix
    Q = at.as_tensor_variable(Q + np.eye(n)*1e-6)  # pequeña corrección para inversa

    # Efectos espaciales
    u = pm.MvNormal("u", mu=at.zeros(n), cov=tau**-2 * at.nlinalg.matrix_inverse(Q), shape=n)

    # Modelo lineal + CAR
    eta = at.dot(X, beta) + u
    p = pm.Deterministic("p", pm.math.sigmoid(eta))

    # Likelihood binario
    y_obs = pm.Bernoulli("y_obs", p=p, observed=y)

    # Muestreo
    trace = pm.sample(2000, tune=1000, target_accept=0.95, progressbar=True)

# =============================================
# 📈 VISUALIZAR RESULTADOS
# =============================================
# Media posterior de la probabilidad
p_pred = trace.posterior['p'].mean(dim=("chain", "draw")).values

# Agregar al GeoDataFrame
gdf['prob_uso_plaguicidas'] = p_pred

# Mapa
import matplotlib.pyplot as plt
gdf.plot(column='prob_uso_plaguicidas', cmap='YlOrRd', legend=True, edgecolor='black')
plt.title("Probabilidad de uso de plaguicidas (modelo CAR)")
plt.axis("off")
plt.show()

print(gdf.columns)

```

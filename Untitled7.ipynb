Construcción del modelo ICAR

```python
with pm.Model() as car_model:
    # Intercepto
    alpha = pm.Normal("alpha", mu=0, sigma=10)

    # Parámetro de dispersión
    tau = pm.HalfCauchy("tau", beta=5)

    # Matriz de adyacencia en formato denso
    W = w.toarray()
    D = np.diag(W.sum(axis=1))
    L = D - W  # Laplaciana

    # Para evitar que la matriz no sea invertible (ICAR)
    # hacemos un pequeño truco: eliminar la última fila y columna
    L_reduced = L[:-1, :-1]

    # Vector de efectos espaciales (n-1)
    phi_reduced = pm.MvNormal("phi_reduced", mu=at.zeros(n - 1),
                              cov=at.slinalg.matrix_inverse(tau * L_reduced),
                              shape=(n - 1))

    # Rellenamos el último valor con 0 para fijar la suma a cero (identificabilidad)
    phi = at.concatenate([phi_reduced, at.zeros(1)])

    # Media
    mu = alpha + phi

    # Ruido
    sigma = pm.HalfNormal("sigma", sigma=10)

    # Observaciones
    y_obs = pm.Normal("y_obs", mu=mu, sigma=sigma, observed=y)

    # Muestreo
    trace = pm.sample(2000, tune=1000, target_accept=0.95, return_inferencedata=True)
```
